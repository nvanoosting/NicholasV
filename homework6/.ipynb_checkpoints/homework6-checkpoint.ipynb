```python
# --- NumPy Arrays ---

list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)

print(arr * 4,
    np.mean(arr),
    np.median(arr),
    f"Max : {arr[np.argmax(arr)]}, Min : {arr[np.argmin(arr)]}",
    np.min(arr), # These are different from argmin/argmax because they print the value of the max and min, while argmin/argmax print their indexes
    np.max(arr)
)

new_arr = []
for i in arr:
    if i > 20:
        new_arr.append(i)
print(np.array(new_arr))

matrix = np.array([
[5, 12, 17, 9],
[8, 21, 14, 3],
[6, 13, 19, 11]
])

print(
    np.shape(matrix),
    matrix[:,2],
    matrix[1,:],
    matrix * 2,
    np.mean(matrix, axis = 0),
    np.mean(matrix),
    np.max(matrix, axis = 1),
    np.min(matrix, axis = 0),
    matrix.flatten(), # removes the nested arrays and makes matrix into a single array
    np.unravel_index(np.argmin(matrix), matrix.shape),
    np.unravel_index(np.argmax(matrix), matrix.shape),
)

new_matrix = []
for i in matrix.flatten():
    if i > 15:
        new_matrix.append(i)
print(np.sort(new_matrix))
        
```

```python
# --- Prime Star Systems ---

arr = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])

def is_prime(value):
    if value < 2:
        return False
    if value == 2:
        return True
    for i in range (2, int(value**0.5) + 1):
        if value % i == 0:
            return False
    return True

def prime_rows(list):
    arr = np.array(list)
    final_arr = []
    for row in arr:
        if any(is_prime(value) for value in row):
            row_list = row.tolist()
            if row_list not in final_arr:
                final_arr.append(row_list)
    return final_arr
```

```python
# --- Checkerboard ---

def checkerboard():
    board = np.zeros((8,8))
    for row in board[::2]:
        for value in row:
            row[1::2] = 1
    for row in board[1::2]:
        for value in row:
            row[::2] = 1
    return board

def reverse_checkerboard():
    board = np.zeros((8,8))
    for row in board[1::2]:
        for value in row:
            row[1::2] = 1
    for row in board[::2]:
        for value in row:
            row[::2] = 1
    return board
```

```python
# --- Expanding Universe ---

def expansion(list, num):
    final_list = []
    arr = np.array(list)
    for value in arr:
        expanded_string = ""
        for char in value:
            expanded_string += char + " " * num
        final_list.append(expanded_string)
    return np.array(final_list)

expansion(["Hello","Goodbye"], 1)
```

```python
# --- Second-Brightest Star

def second_brightest(list):
    second_brightest_stars_per_row = []
    matrix = np.array(list)
    for row in matrix:
        rearranged_row = np.partition(row, -2)
        second_brightest_stars_per_row.append(rearranged_row[-2])
    return np.array(second_brightest_stars_per_row)

print(second_brightest([[1123, 1456, 1789, 1324, 1876],
                        [1567, 1987, 1678, 1405, 1589],
                        [1345, 1654, 1523, 1109, 1923],
                        [1298, 1890, 1367, 1784, 1432],
                        [1823, 1756, 1489, 1672, 1550]]))
```
