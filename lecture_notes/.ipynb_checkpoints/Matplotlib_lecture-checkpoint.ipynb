```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = 1 / np.cos(x)

fig, axes = plt.subplots(2,2, figsize = (10, 4))

axes[0,0] = plt.subplot(x, y1, color = "teal", ls = ":")
axes[0,0] = plt.title("Sine Wave")

axes[0,1] = plt.plot(x, y2, color = "cornflowerblue", ls = "--")
axes[0,1] = plt.title("Cosine Wave", color = "darkblue")

axes[1,0] = plt.plot(x, y3, color = "yellow", ls = "--")
axes[1,0] = plt.title("Tan(x)")

axes[1,1] = plt.plot(x, y4, color = "red", ls = ":")
axes[1,1] = plt.title("Sec(x)")

plt.show()
```

```python
fig, axes = plt.subplots(2,2, figsize = (10, 4))
plt.subplots_adjust(hspace=0.5)

axes[0].plot(x, y1, color = "teal", ls = ":")
axes[0].set_title("Sine Wave", color = "green")

axes[1].plot(x, y2, "r-.")
axes[1].set_title("Cosine Wave", color = "red")

axes[2].plot(x, y3, "--")
axes[2].set_title("Tangent Function", color = "yellow")

axes[3].plot(x, y4)
axes[3].set_title("Secant Function", color = "blue")

plt.show()
```
