```python
class Car:
    def __init__(self, Make, Model):
        self.Make = Make
        self.Model = Model
        self.wheels = 4

    def read_gas(self):
        print(f"{self.Make} {self.Model} has a full tank of gas!")
```

```python
thing1 = Car("Toyota", "Camry")
thing2 = Car("Honda", "Acord")

thing1.wheels
thing1.read_gas()

thing2.wheels
thing2.read_gas()
```

```python
class Rectangle():
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

rect1 = Rectangle(5, 6)
rect1.area()

class Square(Rectangle):
    def __init__(self, side_length):
        super().__init__(side_length, side_length)

square1 = Square(5)
square1.area()
```

```python
class Planet:
    def __init__(self, name, mass, radius):
        self.name = name
        self.mass = mass
        self.radius = radius

    def surface_gravity(self):
        G = 6.67 * (10 ** -11)
        return G * self.mass / (self.radius ** 2)

class Rocky(Planet):
    def __init__(self, name, mass, radius, has_life):
        super().__init__(name, mass, radius)
        self.has_life = has_life

    def has_life(self):
        if self.has_life:
            return f"{self.name} has life!"
        else:
            return f"{self.name} does not have life"

class Gas(Planet):
    def __init__(self, name, mass, radius, num_moons):
        super().__init__(name, mass, radius)
        self.num_moons = num_moons

    def has_x_moons(self):
        return f"{self.name} has {self.num_moons} moons!"

Earth = Rocky("Earth", 20, 4, True)
Neptune = Gas("Neptune", 30, 5, 14)

Earth.has_life
```
