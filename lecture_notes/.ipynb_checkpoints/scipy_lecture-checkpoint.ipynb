```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.tan(x)

plt.scatter(x, y)
plt.title("Tan(x)")
plt.show()
```

```python
df = pd.read_csv("F26_cf_demo_data.csv")

print(df.head(), df.tail())

x = df["x"]
y = df["y"]

plt.plot(x, y, "b+", label = "data")
plt.title("Data to be fit")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

plt.show()
```

```python
def linear(x, m, b):
    return m * x + b
```

```python
popt, pcov = curve_fit(linear, x, y)
```

```python
m_fit, b_fit = popt
print(f"line of best fit: {m_fit:.2f}x + {b_fit:.2f}")
```

```python
plt.plot(x, y, "b+", label = "data")
# y_fit = linear(x, m_fit, b_fit)
plt.plot(x, m_fit*x + b_fit, "r", label = "linear fit")
plt.title("Data to be fit")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

plt.show()
```

```python
#line of best fit
plt.plot(x, linear(x, *popt), color = "red", lavel = "li fit")
plt.plot(x, y, "b+", label = "data")
plt.legend()
plt.show()
```

```python
perr = np.sqrt(np.diag(pcov))
m_err, b_err = perr

print(f"Slope: m = {m_fit:.2f} +- {m_err:.2f}")
```

```python
residuals = y - linear(x, *popt)
chi2 = np.sum(residuals**2)
dof = len(x) - len(popt)
chi2_reduced = chi2 / dof\

print(f"reduced chi squared = {chi2_reduced:.3f}")
```
