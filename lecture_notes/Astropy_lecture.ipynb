```python
from astropy import units as u

distance = 1e5 * u.km
velocity = 20 * u.km / u.s
time = 10 * u.s
```

```python
velocity.unit
velocity.value
```

```python
distance_au = distance.to(u.au)
velocity_mpers = velocity.to(u.m / u.s)
time_minutes = time.to(u.min)
```

```python
from astropy import constants as const

speed_of_light = const.c

def energy_calculate(mass):
    m = mass.to(u.kg)
    return m * speed_of_light ** 2
```
