```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.random.rand(50)
y = np.random.rand(50)
plt.scatter(x,y)
plt.title("ScatterPlot")
plt.show()
```

```python
names = ["A","B","C","D"]
values = [1,2,3,4]
plt.bar(names, values)
plt.title("Bar Chart")
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 200)
y1 = np.sin(x)
y2 = np.sin(2 * x)
y3 = np.sin(3 * x)

fig, axes = plt.subplots(1, 3)
plt.subplots_adjust(wspace=0.5) 

axes[0].plot(x, y1)
axes[0].set_title("Function of sin(x)")
axes[0].set_xlabel("x")

axes[1].plot(x, y2)
axes[1].set_title("Function of sin(2x)")
axes[1].set_xlabel("x")

axes[2].plot(x, y3)
axes[2].set_title("Function of sin(3x)")
axes[2].set_xlabel("x")

plt.show()
```

```python
exponential_values = [1, 2, 3]
x = np.linspace(0, 5, 100)

for a in exponential_values:
    plt.plot(x, x ** a, label=f"x^{a}")

plt.legend()
plt.title("Function of x^a")
plt.xlabel("x")
plt.ylabel("x^a")
plt.show()
```
