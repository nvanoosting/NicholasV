```python
import pandas as pd
```

```python
# --- Pandas DataFrame ---

df = pd.read_csv("../course_assignments/homework8/state_data.csv")

print(df.shape, # 52 rows, 4 columns
      df.head(),  # First state is Alaska
      df.tail() # Last state is Wyoming
     )
# There are 52 entries in total
# There are more than 50 states listed because it includes Puerto Rico and DC

df["latitude"].median() # 39.4
df["longitude"].median() # -89.1

df["Absolute Value to Median Latitude"] = abs(df["latitude"] - df["latitude"].median())
df.shape # 52 rows, 5 columns

print(
    f"{(df.loc[(df["latitude"] - df["latitude"].median()).abs().idxmin(), "latitude"]) - df["latitude"].median():.2f}",
df.loc[(df["latitude"] - df["latitude"].median()).abs().idxmin(), "state"])

print(
    f"{(df.loc[(df["latitude"] - df["latitude"].median()).abs().idxmax(), "latitude"]) - df["latitude"].median():.2f}",
df.loc[(df["latitude"] - df["latitude"].median()).abs().idxmax(), "state"])

print(
    f"{(df.loc[(df["longitude"] - df["longitude"].median()).abs().idxmin(), "longitude"]) - df["longitude"].median():.2f}",
df.loc[(df["longitude"] - df["longitude"].median()).abs().idxmin(), "state"])


print(
    f"{(df.loc[(df["longitude"] - df["longitude"].median()).abs().idxmax(), "longitude"]) - df["longitude"].median():.2f}",
df.loc[(df["longitude"] - df["longitude"].median()).abs().idxmax(), "state"])

df.agg(["sum", "min"]) # This function can be found under the user defined functions
# This function will allow you to perform multiple statistical funtions like finding the minimum value or median of a data frame
# This would be useful when you have to perform all these functions and want them all in a new data frame
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# --- Seaborn ---

df = sns.load_dataset("planets")
df.dropna(axis=1)

print(df.shape, # 1035 rows, 6 columns
      df.head(),
      df.tail()
     )
# Method is how the planet was discovered
# Number is the number of that planet discovered
# Oribtal period is how long it takes the planet to orbit its star
# Distance is the distance between us and the planet in AU
# Year is the year of discovery

fig, axes = plt.subplots(2, 2, figsize = (10, 4))
plt.subplots_adjust(wspace = 1)

axes[0, 0].scatter(df["orbital_period"], df["mass"])

axes[0, 1].scatter(df["distance"], df["year"])

axes[1, 0].hist(df["orbital_period"])

sns.boxplot(
    data=df,
    x="distance",
    y="method",
    ax=axes[1, 1],
)

plt.show()
```
